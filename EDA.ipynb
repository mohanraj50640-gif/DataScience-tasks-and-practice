```python
# Create new feature: let predict marksperhour based on marks in each subjects and weeklystudyhours

subjects = [
    'geography_score',
    'english_score',
    'biology_score',
    'math_score',
    'history_score',
    'chemistry_score'
]

student['total_marks'] = student[subjects].sum(axis=1)

student['marks_per_hour'] = (
    student['total_marks'] /
    student['weekly_self_study_hours']
)
```
