```python
# to find outliers on numerical terms
score_columns=[ 'math_score', 'history_score', 'physics_score',
       'chemistry_score', 'biology_score', 'english_score', 'geography_score']

df[score_columns].head()
```

```python
# let create boxplot for 7 subjects
plt.figure(figsize=(12,6))
sns.boxplot(data=df[score_columns])
plt.title("Boxplot for 7 subjects")
plt.xticks(rotation=45)
plt.show()

# observe below dots that are potential outliers
```

```python
# let take maths score and apply iqr method to find outliers
math=df['math_score']
Q1=math.quantile(0.25)
Q3=math.quantile(0.75)
IQR=Q3-Q1
lower_bound=Q1-1.5*IQR
upper_bound=Q3+1.5*IQR

print("Q1",Q1)
print("Q3",Q3)
print("IQR",IQR)
print("lower_bound",lower_bound)
print("upper_bound",upper_bound)
```

```python
# outlier for math score
math_outliers=df[(df['math_score']<lower_bound) | (df['math_score']>upper_bound) ]
math_outliers
```

```python
for col in score_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    print("\nSubject:", col)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)
    print("Number of Outliers:", len(outliers))
```

```python
outlier_summary = []

for col in score_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    count = ((df[col] < lower) | (df[col] > upper)).sum()

    outlier_summary.append([
        col,
        Q1,
        Q3,
        IQR,
        lower,
        upper,
        count
    ])

outlier_df = pd.DataFrame(
    outlier_summary,
    columns=[
        'Subject',
        'Q1',
        'Q3',
        'IQR',
        'Lower_Bound',
        'Upper_Bound',
        'Outlier_Count'
    ]
)

print(outlier_df)
```

```python
df_clean = df.copy()

for col in score_columns:

    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df_clean = df_clean[
        (df_clean[col] >= lower) &
        (df_clean[col] <= upper)
    ]
```

```python
# BEFORE
plt.figure(figsize=(12, 6))

sns.boxplot(data=df[score_columns])

plt.title("Student Scores - Before Outlier Removal")
plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(12, 6))

sns.boxplot(data=df_clean[score_columns])

plt.title("Student Scores - After Outlier Removal")
plt.xticks(rotation=45)

plt.show()
```

```python
sns.histplot(df_clean['math_score'],kde=True)

plt.title("Math Score Distribution After Outlier Removal")
plt.xlabel("Math Score")
plt.ylabel("Number of Students")
plt.show()
```
